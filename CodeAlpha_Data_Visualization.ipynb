```python
import matplotlib.pyplot as plt

subjects = ['Python', 'Excel', 'Power BI', 'SQL']
marks = [80, 90, 75, 85]

plt.bar(subjects, marks)
plt.title('Student Performance')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.show()
```

```python
import matplotlib.pyplot as plt

products = ['Laptop', 'Mobile', 'Tablet']
sales = [50, 80, 30]

plt.bar(products, sales)
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
labels = ['Python', 'Excel', 'SQL']
values = [40, 35, 25]

plt.pie(values, labels=labels, autopct='%1.1f%%')
plt.title('Learning Preferences')
plt.show()
```

```python
hours = [1, 2, 3, 4, 5]
marks = [40, 50, 60, 75, 90]

plt.scatter(hours, marks)
plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```

```python
Bar chart: Compares the sales of different products.
Line chart: Sales increase overall from January to April.
Pie chart: Python has the largest share in the example.
Scatter plot: Marks generally increase as study hours increase.
```
